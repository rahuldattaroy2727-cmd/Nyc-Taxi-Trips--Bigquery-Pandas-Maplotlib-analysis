```python
import pandas as pd
Query = ''' SELECT *
 FROM `bigquery-public-data.biglake-public-nyc-taxi-iceberg.public_data.nyc_taxicab_2021` limit 100000'''

df = client.query(Query).to_dataframe()
```

```python
df
```

```python
df = df.replace(0E-9,0)
```

```python
df['trip_distance'] = df['trip_distance'].astype(float)
df['fare_amount'] = df['fare_amount'].astype(float)
df['mta_tax'] = df['mta_tax'].astype(float)
df['tip_amount'] = df['tip_amount'].astype(float)
df['total_amount'] = df['total_amount'].astype(float)
df['extra'] = df['extra'].astype(float)
df['tolls_amount'] = df['tolls_amount'].astype(float)
```

#Data Cleaning and Tansformation

```python
df['trip_duration']=(df['dropoff_datetime'] - df['pickup_datetime']).dt.total_seconds() /3600
#created a trip duration column where duration of travel is presented in hours
```

```python
df['trip_duration'].sort_values(ascending = False).head(10)

#trips with  duration  more than 5 hours seems  a lot suspicious
```

```python
df[df['trip_duration']>5].count()
# here we can see that there are 50 rows where trip duration is more than  5 hours
```

```python
long_trips = df[df['trip_duration']>5]
```

```python
long_trips[['pickup_datetime','dropoff_datetime','trip_duration','trip_distance','fare_amount']]
# from checking the trip duration ,trip distance and fare amount the anomalies are confirmed that these are really bad data
```

```python
df2 = df.copy()
df2 = df2[df2['trip_duration']<=5]

# we are making a safe copy df so that any chages in df2 will not affect the orignal dataset
```

```python
df2
#so df2 is\ cleaned where the trips more than 5 hours has been filtered out
```

```python
df2['trip_duration_minutes'] = df2['trip_duration']*60
# we made a another columns with trip duration in minutes
```

```python
df2['year']=df2['pickup_datetime'].dt.year
```

```python
df2['year'].value_counts()
#so we know in the dataset taxi trips 2021 that we have some years that is outside the intended  year. judging by the number of counts we can safely remove them .
```

```python
df2  = df2[df2['year']==2021] # we are keeping 2021 data
```
